```python
import numpy as np
from mplsoccer import Pitch
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from matplotlib.patches import Arc, Wedge, Rectangle
import cv2
import matplotlib
matplotlib.use('Qt5Agg')
```

# 1) Create soccer pitch

### 1.1) Create full pitch

```python
def visualize_soccer_pitch_full():
    # Pitch size
    width, height = 2100, 1360

    # Create green tones
    grass_dark = np.zeros((height, width, 3), dtype=np.uint8)
    grass_dark[:, :] = (17, 101, 14)

    grass_light = np.zeros((height, width, 3), dtype=np.uint8)
    grass_light[:, :] = (22, 124, 21)

    # Create random variations for a natural grass look
    noise = np.zeros((height, width), dtype=np.uint8)
    cv2.randu(noise, 0, 30)
    grass_dark[:, :, 1] = cv2.add(grass_dark[:, :, 1], noise)
    grass_light[:, :, 1] = cv2.add(grass_light[:, :, 1], noise)

    # Create stripes
    grass_striped = grass_dark.copy()
    stripe_width = 123.5
    center_stripe_width = 124
    
    for i in range(4):
        start = int(i * stripe_width * 2)
        end = int(start + stripe_width)
        grass_striped[:, start:end, :] = grass_light[:, start:end, :]
    
    center_start = int((width - center_stripe_width) // 2)
    grass_striped[:, center_start:center_start+center_stripe_width, :] = grass_light[:, center_start:center_start+center_stripe_width, :]
    
    for i in range(4, 9):
        start = int(i * stripe_width * 2)
        end = int(start + stripe_width)
        grass_striped[:, start:end, :] = grass_light[:, start:end, :]

    plt.imshow(cv2.cvtColor(grass_striped, cv2.COLOR_BGR2RGB))

    # create pitch
    pitch = Pitch(pitch_type="custom", pitch_length=105, pitch_width=68, line_alpha=1, line_color="white", corner_arcs=True, linewidth=1)
    fig, ax = pitch.draw()

    ax.imshow(grass_dark, extent=[-4, 109, -4, 72], aspect='auto')
    ax.imshow(grass_light, extent=[0, 105, 0, 68], aspect='auto')

    # add outside line
    ax.add_patch(plt.Rectangle((0, 0), 105, 68, edgecolor="white", facecolor="none", linewidth=1))

    plt.show()

    # Save
    dpi = 300
    fig.set_size_inches((width+250)/dpi, (height+250)/dpi)
    fig.savefig("src\\assets\\top-view\\pitch_soccer_full.png", dpi=dpi, bbox_inches="tight", pad_inches=0)

    return fig

graphic_soccer_pitch = visualize_soccer_pitch_full()
```

# 3) Create basketball court
