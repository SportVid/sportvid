# 0) Import

```python
import numpy as np
from mplsoccer import Pitch
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
from matplotlib.patches import Arc, Wedge, Rectangle
import cv2
import matplotlib
matplotlib.use('Qt5Agg')
```

# 1) Create soccer pitch

### 1.1) Create full pitch

```python
def visualize_soccer_pitch_full():
    # Pitch size
    width, height = 2100, 1360

    # Create green tones
    grass_dark = np.zeros((height, width, 3), dtype=np.uint8)
    grass_dark[:, :] = (17, 101, 14)

    grass_light = np.zeros((height, width, 3), dtype=np.uint8)
    grass_light[:, :] = (22, 124, 21)

    # Create random variations for a natural grass look
    noise = np.zeros((height, width), dtype=np.uint8)
    cv2.randu(noise, 0, 30)
    grass_dark[:, :, 1] = cv2.add(grass_dark[:, :, 1], noise)
    grass_light[:, :, 1] = cv2.add(grass_light[:, :, 1], noise)

    # Create stripes
    grass_striped = grass_dark.copy()
    stripe_width = 123.5
    center_stripe_width = 124
    
    for i in range(4):
        start = int(i * stripe_width * 2)
        end = int(start + stripe_width)
        grass_striped[:, start:end, :] = grass_light[:, start:end, :]
    
    center_start = int((width - center_stripe_width) // 2)
    grass_striped[:, center_start:center_start+center_stripe_width, :] = grass_light[:, center_start:center_start+center_stripe_width, :]
    
    for i in range(4, 9):
        start = int(i * stripe_width * 2)
        end = int(start + stripe_width)
        grass_striped[:, start:end, :] = grass_light[:, start:end, :]

    plt.imshow(cv2.cvtColor(grass_striped, cv2.COLOR_BGR2RGB))

    # create pitch
    pitch = Pitch(pitch_type="custom", pitch_length=105, pitch_width=68, line_alpha=1, line_color="white", corner_arcs=True, linewidth=1)
    fig, ax = pitch.draw()

    ax.imshow(grass_dark, extent=[-4, 109, -4, 72], aspect='auto')
    ax.imshow(grass_striped, extent=[0, 105, 0, 68], aspect='auto')

    # add outside line
    ax.add_patch(plt.Rectangle((0, 0), 105, 68, edgecolor="white", facecolor="none", linewidth=1))

    plt.show()

    # Save
    dpi = 300
    fig.set_size_inches((width+250)/dpi, (height+250)/dpi)
    fig.savefig("src\\assets\\top-view\\pitch_soccer_full.png", dpi=dpi, bbox_inches="tight", pad_inches=0)

    return fig

graphic_soccer_pitch = visualize_soccer_pitch_full()
```

# 3) Create basketball court

### 3.1) Create court background

```python
def visualize_basketball_court_background():
    # Pitch size
    width, height = 2865, 1524

    # Create brown tones
    wood_colors = [
        (238, 214, 194),  # light
        (238, 214, 194),  # light
        (238, 214, 194),  # light
        (225, 201, 185),   # mid
        (204, 180, 163),   # dark
        (244, 216, 190),  # lighter
    ]

    # Create random variations for a natural wood grain
    wood_texture = np.zeros((height, width, 3), dtype=np.uint8)

    y = 0
    while y < height:
        plank_height = np.random.randint(6, 11)
        x = 0
        while x < width:
            plank_width = np.random.randint(40, 71)
            color = wood_colors[np.random.randint(0, len(wood_colors))]
            wood_texture[y:y+plank_height, x:x+plank_width] = color
            x += plank_width
        y += plank_height

    noise = np.random.randint(0, 30, (height, width, 1), dtype=np.int8)
    wood_texture = np.clip(wood_texture + np.repeat(noise, 3, axis=2), 0, 255).astype(np.uint8)

    plt.figure(figsize=(10, 6))
    plt.imshow(wood_texture)
    plt.axis("off")

    return plt

graphic_basketball_court_background = visualize_basketball_court_background()
graphic_basketball_court_background.savefig("src\\assets\\top-view\\background_basketball.png")
```

```python
# 5) Check actual PNG dimensions and calculate correct widthRel/heightRel
from PIL import Image
import os

def check_pitch_dimensions():
    """Calculate correct widthRel and heightRel for each area size"""
    
    # Define the playing field dimensions
    pitch_sizes = {
        "full": {"width": 2100, "height": 1360},
        "half": {"width": 1050, "height": 1360},
        "penToPen": {"width": 1440, "height": 1360},  # 72m length converted
        "doublePen": {"width": 660, "height": 806},
    }
    
    # Define PNG file paths
    png_files = {
        "full": "src/assets/top-view/pitch_soccer_full.png",
        "half": "src/assets/top-view/pitch_soccer_half.png",
        "penToPen": "src/assets/top-view/pitch_soccer_pen_to_pen.png",
        "doublePen": "src/assets/top-view/pitch_soccer_double_pen.png",
    }
    
    print("=" * 70)
    print("PITCH DIMENSION ANALYSIS")
    print("=" * 70)
    
    for size_name, pitch_dim in pitch_sizes.items():
        png_path = png_files[size_name]
        
        if os.path.exists(png_path):
            img = Image.open(png_path)
            png_width, png_height = img.size
            
            # Calculate widthRel and heightRel
            width_rel = pitch_dim["width"] / png_width
            height_rel = pitch_dim["height"] / png_height
            
            print(f"\n{size_name.upper()}:")
            print(f"  Playing Field:  {pitch_dim['width']} × {pitch_dim['height']}")
            print(f"  PNG Actual:     {png_width} × {png_height}")
            print(f"  → widthRel:  {pitch_dim['width']} / {png_width} = {width_rel:.6f}")
            print(f"  → heightRel: {pitch_dim['height']} / {png_height} = {height_rel:.6f}")
            print(f"  → Code: widthRel: {pitch_dim['width']} / {png_width},")
            print(f"          heightRel: {pitch_dim['height']} / {png_height},")
        else:
            print(f"\n{size_name.upper()}: ⚠️ PNG not found at {png_path}")

check_pitch_dimensions()
```
